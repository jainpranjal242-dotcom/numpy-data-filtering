# NumPy Data Filtering
**VEDA Technology – AI & ML Track | Level 1 – Day 13**

```python
import numpy as np
data = np.array([25, 40, 55, 60, 72, 85, 90, 45, 30, 65])
print('Original Data:', data)
```

```python
print('Above 50:', data[data > 50])
print('Below 50:', data[data < 50])
print('Equal to 60:', data[data == 60])
print('Above 70:', data[data > 70])
print('Greater than or equal to 60:', data[data >= 60])
```

## Explanation
Boolean indexing selects NumPy array elements using True/False conditions. Example: `data[data > 50]` returns only values greater than 50.

## Interview Questions
1. What is Boolean indexing? Selecting elements using a Boolean condition.
2. How do you filter a NumPy array? Use a condition inside array indexing, e.g. `data[data > 50]`.
3. What is a Boolean array? An array containing True/False values.
